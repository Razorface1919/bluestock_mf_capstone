```python
# NAV Trend Analysis (Fixed Column Name)
import plotly.express as px

query = """
    SELECT f.date as nav_date, f.nav, d.scheme_name 
    FROM fact_nav f
    JOIN dim_fund d ON f.amfi_code = d.amfi_code
"""
# We use parse_dates=['nav_date'] because we aliased f.date AS nav_date in the query above
df_nav = pd.read_sql(query, conn, parse_dates=['nav_date'])

fig = px.line(df_nav, x='nav_date', y='nav', color='scheme_name', 
              title='Daily NAV Trend for 40 Schemes (2022-2026)',
              labels={'nav_date': 'Date', 'nav': 'Net Asset Value (₹)'})

# Highlight 1: COVID Recovery (Early 2022)
fig.add_vrect(x0="2022-01-01", x1="2022-06-30", 
              fillcolor="blue", opacity=0.1, line_width=0, annotation_text="COVID Recovery")

# Highlight 2: 2023 Rally
fig.add_vrect(x0="2023-03-01", x1="2023-12-31", 
              fillcolor="green", opacity=0.1, line_width=0, annotation_text="2023 Rally")

# Highlight 3: 2024 Corrections
fig.add_vrect(x0="2024-05-01", x1="2024-07-31", 
              fillcolor="red", opacity=0.1, line_width=0, annotation_text="2024 Correction")

fig.update_layout(showlegend=False) # Hiding legend for 40 funds to avoid clutter
fig.show()
fig.write_image(r'reports\nav_trend.png')
```

```python
# AUM Growth Bar Chart
df_aum = pd.read_csv(r'data\raw\03_aum_by_fund_house.csv', parse_dates=['date'])
df_aum['year'] = df_aum['date'].dt.year

# Filtering for top 5 fund houses to keep the chart readable
top_amcs = df_aum.groupby('fund_house')['aum_lakh_crore'].max().nlargest(5).index
df_aum_top = df_aum[df_aum['fund_house'].isin(top_amcs)]

plt.figure(figsize=(14, 7))
ax = sns.barplot(data=df_aum_top, x='year', y='aum_lakh_crore', hue='fund_house', ci=None)
plt.title('AUM Growth by Top Fund Houses (2022-2025)', fontsize=16)
plt.ylabel('AUM (₹ Lakh Crore)')
plt.xlabel('Year')

# Annotating SBI's Dominance in 2025
plt.annotate('SBI Dominance: ₹12.5L Cr', 
             xy=(3.1, 12.5), xytext=(2.5, 13),
             arrowprops=dict(facecolor='black', shrink=0.05))
plt.tight_layout()
plt.savefig(r'reports\aum_growth.png')
plt.show()
```

```python
# SIP Inflow Time-Series
df_sip = pd.read_csv(r'data\raw\04_monthly_sip_inflows.csv')
df_sip['month'] = pd.to_datetime(df_sip['month'])

fig = px.line(df_sip, x='month', y='sip_inflow_crore', markers=True,
              title='Monthly SIP Inflows (Jan 2022 - Dec 2025)',
              labels={'month': 'Month', 'sip_inflow_crore': 'SIP Inflow (₹ Crore)'})

# Annotating the Dec 2025 All-Time High
fig.add_annotation(x='2025-12-01', y=31002,
            text="All-Time High: ₹31,002 Cr",
            showarrow=True, arrowhead=1, ax=-50, ay=-40)

fig.show()
```

```python
# Category Inflow Heatmap
df_cat = pd.read_csv(r'data\raw\05_category_inflows.csv')
pivot_cat = df_cat.pivot(index='category', columns='month', values='net_inflow_crore')

plt.figure(figsize=(16, 8))
sns.heatmap(pivot_cat, cmap='RdYlGn', center=0, annot=False, linewidths=.5)
plt.title('Net Inflows by Fund Category (FY 2024-25)', fontsize=16)
plt.ylabel('Fund Category')
plt.xlabel('Month')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig(r'reports\category_heatmap.png')
plt.show()
```

```python
# Investor Demographics
df_txn = pd.read_sql("SELECT * FROM fact_transactions", conn)
sip_txns = df_txn[df_txn['transaction_type'].str.upper() == 'SIP']

fig, axes = plt.subplots(1, 3, figsize=(20, 6))

# 1. Pie Chart: Age Distribution
age_counts = df_txn['age_group'].value_counts()
axes[0].pie(age_counts, labels=age_counts.index, autopct='%1.1f%%', startangle=90, colors=sns.color_palette("Set3"))
axes[0].set_title('Investor Age Group Distribution')

# 2. Box Plot: SIP Amount by Age
sns.boxplot(data=sip_txns, x='age_group', y='amount_inr', ax=axes[1], order=['18-25', '26-35', '36-45', '46-55', '56+'])
axes[1].set_title('SIP Amount Distribution by Age Group')
axes[1].set_ylabel('SIP Amount (₹)')

# 3. Pie Chart: Gender Split
gender_counts = df_txn['gender'].value_counts()
axes[2].pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%', startangle=90, colors=['#ff9999','#66b3ff'])
axes[2].set_title('Investor Gender Split')

plt.tight_layout()
plt.savefig(r'reports\demographics.png')
plt.show()
```

```python
# Geographic Distribution
fig, axes = plt.subplots(1, 2, figsize=(18, 7))

# Horizontal Bar: SIP by State
state_sip = sip_txns.groupby('state')['amount_inr'].sum().sort_values(ascending=False)
sns.barplot(x=state_sip.values, y=state_sip.index, ax=axes[0], palette='viridis')
axes[0].set_title('Total SIP Inflows by State')
axes[0].set_xlabel('Total Amount (₹)')

# Pie Chart: T30 vs B30
tier_counts = df_txn['city_tier'].value_counts()
axes[1].pie(tier_counts, labels=tier_counts.index, autopct='%1.1f%%', colors=['#ff9999','#66b3ff'])
axes[1].set_title('City Tier Breakdown (T30 vs B30)')

plt.tight_layout()
plt.savefig(r'reports\geo_distribution.png')
plt.show()
```

```python
# Correlation Matrix
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

query_corr = """
    SELECT f.date as nav_date, f.nav, d.amfi_code
    FROM fact_nav f
    JOIN dim_fund d ON f.amfi_code = d.amfi_code
    WHERE d.amfi_code IN (
        SELECT amfi_code FROM dim_fund LIMIT 10
    )
"""
df_raw = pd.read_sql(query_corr, conn, parse_dates=['nav_date'])

# --- THE FIX: Force NAV to be a numeric float ---
df_raw['nav'] = pd.to_numeric(df_raw['nav'], errors='coerce')

# Pivot to get Dates as rows and AMFI codes as columns
pivot_nav = df_raw.pivot(index='nav_date', columns='amfi_code', values='nav')

# Calculate Daily Returns 
pivot_returns = pivot_nav.pct_change().iloc[1:]

# Compute the correlation matrix
corr_matrix = pivot_returns.corr()

# Plot the heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, 
            annot=True,          
            cmap='coolwarm',     
            fmt=".2f",           
            linewidths=0.5,      
            vmin=0.5, vmax=1.0)  

plt.title('Daily Return Correlation Matrix (10 Selected Funds)', fontsize=14)
plt.ylabel('AMFI Code')
plt.xlabel('AMFI Code')
plt.tight_layout()
plt.savefig(r'reports\correlation_matrix.png')
plt.show()
```

```python
# Sector Allocation Donut
df_holdings = pd.read_csv(r'data\raw\09_portfolio_holdings.csv')
sector_weights = df_holdings.groupby('sector')['weight_pct'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 10))
plt.pie(sector_weights, labels=sector_weights.index, autopct='%1.1f%%', pctdistance=0.85, colors=sns.color_palette('pastel'))

centre_circle = plt.Circle((0,0), 0.70, fc='white')
fig = plt.gcf()
fig.gca().add_artist(centre_circle)

plt.title('Top 10 Sector Allocations Across Equity Funds')
plt.savefig(r'reports\sector_donut.png')
plt.show()
```

# Executive Summary: Top 10 EDA Findings

1. **The Post-2024 Rebound (NAV Resilience):** Even with the sharp 2024 market corrections, all 40 tracked schemes bounced back quickly. The momentum from the 2023 bull run essentially acted as a cushion, keeping the long-term trajectory pointing up.
2. **SBI is in a League of Its Own (AMC Dominance):** SBI Mutual Fund is completely dominating the AUM charts, officially crossing the massive ₹12.5 Lakh Crore mark in 2025. 
3. **Retail Investors Aren't Flinching (Retail Confidence):** December 2025 was a record-breaker. We hit an all-time high of ₹31,002 Crore in SIP inflows, proving that retail investors are consistently pouring money into the market regardless of short-term noise.
4. **The Liquid Fund Surprise (Category Inflows):** Forget the equity hype—when we look at the actual net inflows for FY 2024-25, Liquid funds absolutely stole the show. Traditional equity categories (Large, Mid, Small Cap) were surprisingly sluggish by comparison.
5. **The "Who" and "How Much" (Demographics):** The 26-35 age bracket makes up the biggest chunk of our investor base (41.1%). But the most interesting takeaway? No matter the age group, the median SIP investment stays incredibly flat at around ₹5,000.
6. **A Massive Gender Gap:** The data shows a heavy demographic skew, with 66.5% male investors compared to just 33.5% female. This highlights a massive untapped market for targeted financial inclusion campaigns.
7. **New Geographical Heavyweights:** This was the biggest curveball in the dataset. Instead of traditional financial hubs like Maharashtra or Gujarat leading the pack, states like Madhya Pradesh and Punjab actually dominate the total SIP inflow volume. 
8. **T30 Still Rules (B30 Penetration):** Capital from "Beyond Top 30" (B30) cities only makes up 33.7% of the pie. The major T30 cities are still doing the heavy lifting, meaning AMC expansion into smaller, tier-2/3 cities has a lot of room to grow.
9. **Folios Are Doubling (Industry Growth):** The sheer volume of active accounts is exploding. Total industry folios went from 13.26 Crore in Jan 2022 to a staggering 26.12 Crore by Dec 2025.
10. **The Correlation Anomaly:** Even though equity portfolios are extremely heavy on Banking (22.0%) and IT (15.3%), our 10 specifically selected funds showed near-zero correlation in their daily returns. They are moving highly independently of one another.
